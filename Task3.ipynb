## Key Insights

1. Movies make up a larger portion of the Netflix dataset than TV Shows.

2. The amount of Netflix content has changed significantly across different release years.

3. The United States is one of the major countries producing Netflix content.

4. International Movies, Dramas, and Comedies are among the most common content categories.

5. Netflix contains content with a wide range of ratings for different audiences.

```python
df = pd.read_csv('/content/netflix_data/netflix_titles.csv')
```

```python
df.columns
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='type')

plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

df['release_year'].value_counts().sort_index().plot()

plt.title('Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

top_countries = df['country'].value_counts().head(10)

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

top_genres = df['listed_in'].value_counts().head(10)

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

top_ratings = df['rating'].value_counts().head(10)

sns.barplot(x=top_ratings.index, y=top_ratings.values)

plt.title('Top 10 Netflix Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)
plt.show()
```
